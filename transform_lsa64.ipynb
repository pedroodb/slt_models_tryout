```python
import os


path = "/mnt/sda2/datasets/isolated-cont-sl/lsa64/all_cut"
files = [
    file for file in os.listdir(path) if file.endswith(".json") and len(file) == 16
]
```

```python
import json
import numpy as np


for file in files:
    keypoints = json.load(open(os.path.join(path, file)))

    pose = np.array(
        list(
            [[keypoint["x"], keypoint["y"], keypoint["z"]] for keypoint in frame]
            for frame in keypoints["pose"]
        )
    )
    right = np.array(
        list(
            [[keypoint["x"], keypoint["y"], keypoint["z"]] for keypoint in frame]
            for frame in keypoints["right"]
        )
    )
    left = np.array(
        list(
            [[keypoint["x"], keypoint["y"], keypoint["z"]] for keypoint in frame]
            for frame in keypoints["left"]
        )
    )
    # if number of frame is not the same, pad the shorter one with zeros
    max_len = max(len(pose), len(right), len(left))
    if len(pose) < max_len:
        pose = np.concatenate([pose, np.zeros((max_len - len(pose), 33, 3))], axis=0)
    if len(right) < max_len:
        right = np.concatenate([right, np.zeros((max_len - len(right), 21, 3))], axis=0)
    if len(left) < max_len:
        left = np.concatenate([left, np.zeros((max_len - len(left), 21, 3))], axis=0)

    keypoints = np.concatenate([pose, right, left], axis=1)
    # store the keypoints in a folder named lsa64
    np.save(os.path.join("lsa64", file[:-5]), keypoints)
```
